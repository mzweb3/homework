## ۱. مرتب‌سازی درجی (Insertion Sort)
از عنصر دوم شروع می‌کنیم و هر عنصر را در جای درست خودش در بخش مرتب‌شده‌ی سمت چپ درج می‌کنیم

- بدترین حالت: O(n²)
- بهترین حالت (آرایه‌ی مرتب): O(n)
- حافظه: O(1)
- پایدار (Stable): بله

```python
def insertion_sort(arr):
    n = 0
    for _ in arr:
        n += 1

    i = 1
    while i < n:
        key = arr[i]        
        j = i - 1
        while j >= 0 and arr[j] > key:
            arr[j + 1] = arr[j]
            j -= 1

        arr[j + 1] = key    
        i += 1

    return arr


numbers = [64, 25, 12, 22, 11]
insertion_sort(numbers)
print(numbers)
```

## ۲. برج هانوی (Tower of Hanoi)
برای جابه‌جایی `n` دیسک از مبدأ به مقصد:

1. `n-1` دیسک بالایی را به میله‌ی کمکی ببر.
2. بزرگ‌ترین دیسک را به مقصد ببر.
3. `n-1` دیسک را از میله‌ی کمکی روی دیسک بزرگ در مقصد بگذار.

- تعداد حرکات: 2ⁿ − 1
- پیچیدگی زمانی: O(2ⁿ)
- پیچیدگی حافظه: O(n) به دلیل عمق بازگشت

```python
def hanoi(n, source, target, auxiliary, moves):
    if n == 0:
        return moves
    moves = hanoi(n - 1, source, auxiliary, target, moves)
    moves = moves + [(n, source, target)]
    moves = hanoi(n - 1, auxiliary, target, source, moves)
    return moves

result = hanoi(3, "A", "C", "B", [])
step = 1
for disk, src, dst in result:
    print("حرکت", step, ": دیسک", disk, "از", src, "به", dst)
    step += 1
```
