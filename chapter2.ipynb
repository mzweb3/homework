## ۱. بازگشت و رابطهٔ بازگشتی

**الگوریتم بازگشتی** الگوریتمی است که برای حل یک مسئله، تابع خودش را با ورودی کوچک‌تر فراخوانی می‌کند.

هر تابع بازگشتی باید دو بخش اصلی داشته باشد:

1. **شرط پایان (Base Case):** حالتی که تابع دیگر خودش را صدا نمی‌زند.
2. **گام بازگشتی (Recursive Step):** بخشی که مسئله را کوچک‌تر می‌کند و دوباره تابع را فراخوانی می‌کند.

اگر شرط پایان وجود نداشته باشد یا ورودی به آن نزدیک نشود، فراخوانی‌ها ادامه پیدا می‌کنند تا برنامه با خطای عمق بازگشت متوقف شود.

**رابطهٔ بازگشتی زمان اجرا** نشان می‌دهد زمان اجرای مسئلهٔ اندازهٔ `n` از زمان اجرای زیرمسئله‌ها و کارهای خارج از فراخوانی‌ها چگونه به دست می‌آید.

برای نمونه، اگر فقط یک زیرمسئله با اندازهٔ نصف داشته باشیم و در هر مرحله یک مقدار ثابت کار انجام دهیم:

$$

T(n)=T(n/2)+\Theta(1)

$$

در این دفترچه، `T(n)` زمان اجرای الگوریتم برای ورودی به اندازهٔ `n` است. نمادهای مهم:

- $O(g(n))$: کران بالا برای رشد تابع.
- $\Omega(g(n))$: کران پایین.
- $\Theta(g(n))$: کران دقیق از نظر مرتبهٔ رشد؛ هم کران بالا و هم پایین.

در تحلیل‌ها، هزینهٔ هر مقایسه، انتساب یا عملیات ساده را ثابت در نظر می‌گیریم.

## ۲. قضیهٔ مستر (Master Theorem)

قضیهٔ مستر برای حل گروهی از رابطه‌های بازگشتی به شکل زیر استفاده می‌شود:

$$

T(n)=aT\left(\frac{n}{b}\right)+f(n)

$$

معنی پارامترها:

- `a`: تعداد زیرمسئله‌های بازگشتی
- `b`: ضریب تقسیم اندازهٔ ورودی هر زیرمسئله
- `f(n)`: هزینهٔ کارهای خارج از فراخوانی‌های بازگشتی

ابتدا محاسبه می‌کنیم:

$$

d=\log_b a

$$

سپس تابع `f(n)` را با $n^d$ مقایسه می‌کنیم.

### حالت اول: هزینهٔ بیرون از بازگشت کوچک‌تر است

اگر برای یک $\varepsilon>0$ داشته باشیم:

$$

f(n)=O(n^{d-\varepsilon})

$$

آنگاه:

$$

\boxed{T(n)=\Theta(n^d)}

$$

### حالت دوم: دو بخش هم‌مرتبه‌اند

اگر:

$$

f(n)=\Theta(n^d)

$$

آنگاه:

$$

\boxed{T(n)=\Theta(n^d\log n)}

$$

### حالت سوم: هزینهٔ بیرون از بازگشت بزرگ‌تر است

اگر برای یک $\varepsilon>0$:

$$

f(n)=\Omega(n^{d+\varepsilon})

$$

و شرط منظم بودن برقرار باشد، آنگاه:

$$

\boxed{T(n)=\Theta(f(n))}

$$

شرط منظم بودن در این حالت:

$$

a f(n/b)\leq c f(n),\qquad 0<c<1

$$

### نمونهٔ کوتاه از حالت سوم

رابطهٔ زیر را در نظر بگیر:

$$

T(n)=2T(n/2)+n^2

$$

در اینجا $a=2$، $b=2$ و $f(n)=n^2$ است. پس:

$$

d=\log_2 2=1\quad\Rightarrow\quad n^d=n

$$

چون $n^2$ از $n$ به‌طور چندجمله‌ای بزرگ‌تر است، شرط مقایسه برقرار است. شرط منظم بودن را هم بررسی می‌کنیم:

$$

2f(n/2)=2(n/2)^2=n^2/2\leq (1/2)n^2

$$

پس حالت سوم برقرار است و جواب:

$$

T(n)=\Theta(n^2)

$$

### ترتیب حل در سؤال امتحان

1. `a` و `b` را از رابطه پیدا کن.
2. $\log_b a$ را حساب کن.
3. تابع `f(n)` را مشخص کن.
4. `f(n)` را با $n^{\log_b a}$ مقایسه کن.
5. حالت مناسب را انتخاب کن؛ در حالت سوم شرط منظم بودن را هم بررسی کن.

**نکته:** قضیهٔ مستر مستقیماً برای رابطه‌هایی مثل $T(n)=T(n-1)+1$ مناسب نیست؛ چون اندازهٔ ورودی در آن‌ها یک واحد کم می‌شود، نه اینکه بر یک عدد ثابت تقسیم شود. برای این رابطه‌ها از درخت بازگشتی یا گسترش رابطه استفاده می‌کنیم.

در مثال‌های تقسیم و حل، برای ساده‌تر شدن محاسبات فرض می‌کنیم `n` توانی از ۲ است. این فرض نتیجهٔ مرتبهٔ رشد را تغییر نمی‌دهد.

## ۳. جست‌وجوی دودویی (Binary Search)

**کاربرد:** پیدا کردن یک مقدار در آرایهٔ مرتب‌شده.

در هر مرحله، عضو وسط محدوده را بررسی می‌کنیم:

- اگر عضو وسط برابر مقدار موردنظر باشد، اندیس آن را برمی‌گردانیم.
- اگر مقدار موردنظر کوچک‌تر باشد، نیمهٔ چپ را جست‌وجو می‌کنیم.
- اگر بزرگ‌تر باشد، نیمهٔ راست را جست‌وجو می‌کنیم.
- اگر محدوده خالی شود، مقدار `-1` برمی‌گردانیم؛ یعنی عدد پیدا نشده است.

**شرط مهم:** آرایه باید صعودی مرتب شده باشد. کد زیر برای آرایهٔ صعودی نوشته شده است.

### رابطهٔ بازگشتی

در هر مرحله فقط یک زیرمسئله با تقریباً نصف اندازه باقی می‌ماند و مقایسه‌ها زمان ثابت دارند:

$$

T(n)=T(n/2)+\Theta(1)

$$

### حل با قضیهٔ مستر

$$

a=1,\qquad b=2,\qquad f(n)=\Theta(1)

$$

$$

\log_2 1=0\quad\Rightarrow\quad n^{\log_2 1}=n^0=1

$$

پس `f(n)` با $n^0$ هم‌مرتبه است؛ بنابراین حالت دوم مستر برقرار است:

$$

\boxed{T(n)=\Theta(\log n)}

$$

### چرا لگاریتمی است؟

بعد از `k` مرحله، اندازهٔ محدوده تقریباً برابر $n/2^k$ می‌شود. در انتها به یک عضو می‌رسیم:

$$

\frac{n}{2^k}=1\quad\Rightarrow\quad n=2^k

$$

با گرفتن لگاریتم پایهٔ ۲:

$$

k=\log_2 n

$$

پس تعداد مراحل متناسب با $\log n$ است.

### شبه‌کد

```text
BINARY_SEARCH(A, target, left, right)

    if left > right
        return -1

    mid = (left + right) // 2

    if A[mid] == target
        return mid

    if target < A[mid]
        return BINARY_SEARCH(A, target, left, mid - 1)

    return BINARY_SEARCH(A, target, mid + 1, right)
```

### کد پایتون

````python
def binary_search(arr, target, left, right):
    if left > right:
        return -1
    mid = (left + right) // 2
    if arr[mid] == target:
        return mid
    elif target < arr[mid]:
        return binary_search(arr, target, left, mid - 1)
    else:
        return binary_search(arr, target, mid + 1, right)
numbers = [2, 5, 8, 12, 16, 23, 38, 45]
print("اندیس عدد 23:", binary_search(numbers, 23, 0, len(numbers) - 1))
print("اندیس عدد 20:", binary_search(numbers, 20, 0, len(numbers) - 1))
````

**خروجی مورد انتظار:**

```text
اندیس عدد 23: 5
اندیس عدد 20: -1
```

**پیچیدگی زمانی:** در بدترین حالت $\Theta(\log n)$ است. اگر همان عضو وسط در اولین مقایسه پیدا شود، زمان بهترین حالت $\Theta(1)$ خواهد بود.

**پیچیدگی فضایی:** نسخهٔ بازگشتی به دلیل پشتهٔ فراخوانی‌ها $O(\log n)$ فضا مصرف می‌کند.

## ۴. پیدا کردن بیشترین مقدار به روش بازگشتی (Recursive Maximum)

**کاربرد:** پیدا کردن بزرگ‌ترین عدد آرایه، بدون استفاده از تابع آمادهٔ `max`.

آرایه را به دو قسمت تقسیم می‌کنیم. بیشترین مقدار هر قسمت را به‌صورت بازگشتی پیدا می‌کنیم و در پایان دو جواب را با هم مقایسه می‌کنیم.

### رابطهٔ بازگشتی

دو زیرمسئله با اندازهٔ تقریباً نصف داریم و مقایسهٔ دو جواب زمان ثابت می‌گیرد:

$$

T(n)=2T(n/2)+\Theta(1)

$$

### حل با قضیهٔ مستر

$$

a=2,\qquad b=2,\qquad f(n)=\Theta(1)

$$

$$

\log_2 2=1\quad\Rightarrow\quad n^{\log_2 2}=n

$$

چون:

$$

f(n)=\Theta(1)=O(n^{1-1})

$$

حالت اول مستر برقرار است:

$$

\boxed{T(n)=\Theta(n)}

$$

### توضیح با درخت بازگشتی

در سطح صفر یک مسئله داریم؛ در سطح بعد ۲ مسئله، سپس ۴ مسئله و به همین ترتیب. در انتها `n` زیرمسئلهٔ تک‌عضوی داریم. درخت دودویی کامل با `n` برگ، در مجموع `2n-1` گره دارد. در هر گره فقط کار ثابتی انجام می‌شود؛ بنابراین:

$$

T(n)=\Theta(2n-1)=\boxed{\Theta(n)}

$$

### شبه‌کد

```text
FIND_MAX(A, left, right)

    if left == right
        return A[left]

    mid = (left + right) // 2

    max_left = FIND_MAX(A, left, mid)
    max_right = FIND_MAX(A, mid + 1, right)

    if max_left > max_right
        return max_left
    else
        return max_right
```

### کد پایتون

````python
def find_max_part(arr, left, right):
    if left == right:
        return arr[left]

    mid = (left + right) // 2

    max_left = find_max_part(arr, left, mid)
    max_right = find_max_part(arr, mid + 1, right)

    if max_left > max_right:
        return max_left
    else:
        return max_right


def recursive_max(arr):
    if len(arr) == 0:
        raise ValueError("آرایه نباید خالی باشد.")

    return find_max_part(arr, 0, len(arr) - 1)


numbers = [12, 45, 7, 89, 23, 56, 10, 34]
print("بزرگ‌ترین عدد:", recursive_max(numbers))
````

**خروجی مورد انتظار:**

```text
بزرگ‌ترین عدد: 89
```

**پیچیدگی زمانی:** $\Theta(n)$، چون در مجموع تعداد گره‌های درخت متناسب با تعداد عناصر آرایه است.

**پیچیدگی فضایی:** عمق درخت بازگشتی $O(\log n)$ است؛ بنابراین فضای پشتهٔ فراخوانی‌ها $O(\log n)$ خواهد بود.

## ۵. مرتب‌سازی ادغامی (Merge Sort)

**کاربرد:** مرتب‌کردن عناصر آرایه از کوچک به بزرگ.

مرتب‌سازی ادغامی دو بخش دارد:

1. **تقسیم (Divide):** آرایه را به دو نیمه تقسیم می‌کنیم تا هر قسمت فقط یک عضو داشته باشد.
2. **ادغام (Merge):** دو قسمت مرتب را با مقایسهٔ عناصرشان به یک قسمت مرتب تبدیل می‌کنیم.

### رابطهٔ بازگشتی

دو زیرمسئله با اندازهٔ نصف داریم. ادغام دو نیمه نیز در مجموع متناسب با تعداد عناصر زمان می‌برد:

$$

T(n)=2T(n/2)+\Theta(n)

$$

### حل با قضیهٔ مستر

$$

a=2,\qquad b=2,\qquad f(n)=\Theta(n)

$$

$$

\log_2 2=1\quad\Rightarrow\quad n^{\log_2 2}=n

$$

پس `f(n)` با $n^{\log_2 2}$ برابر است و حالت دوم مستر برقرار می‌شود:

$$

\boxed{T(n)=\Theta(n\log n)}

$$

### چرا $n\log n$؟

تعداد سطح‌های تقسیم تقریباً $\log_2 n$ است. در هر سطح، ادغام‌ها در مجموع همهٔ `n` عنصر را بررسی می‌کنند؛ پس هزینهٔ هر سطح $\Theta(n)$ و تعداد سطح‌ها $\Theta(\log n)$ است:

$$

T(n)=\Theta(n)\times\Theta(\log n)=\boxed{\Theta(n\log n)}

$$

### شبه‌کد

```text
MERGE_SORT(A, left, right)

    if left >= right
        return

    mid = (left + right) // 2

    MERGE_SORT(A, left, mid)
    MERGE_SORT(A, mid + 1, right)

    MERGE(A, left, mid, right)


MERGE(A, left, mid, right)

    i = left
    j = mid + 1
    temp = empty list

    while i <= mid and j <= right
        if A[i] <= A[j]
            add A[i] to temp
            i = i + 1
        else
            add A[j] to temp
            j = j + 1

    add the remaining elements of the left half to temp
    add the remaining elements of the right half to temp

    copy temp back into A[left ... right]
```

### کد پایتون

````python
def merge(arr, left, mid, right):
    temp = []
    i = left
    j = mid + 1
    while i <= mid and j <= right:
        if arr[i] <= arr[j]:
            temp.append(arr[i])
            i += 1
        else:
            temp.append(arr[j])
            j += 1
    while i <= mid:
        temp.append(arr[i])
        i += 1
    while j <= right:
        temp.append(arr[j])
        j += 1
    for k in range(len(temp)):
        arr[left + k] = temp[k]
def merge_sort(arr, left, right):
    if left < right:
        mid = (left + right) // 2

        merge_sort(arr, left, mid)
        merge_sort(arr, mid + 1, right)

        merge(arr, left, mid, right)

    return arr
numbers = [38, 12, 27, 43, 9, 31, 18, 25]
print("آرایهٔ مرتب‌شده:", merge_sort(numbers, 0, len(numbers) - 1))
other_numbers = [4, -2, 4, 0, 7, -5]
print("نمونهٔ دوم:", merge_sort(other_numbers, 0, len(other_numbers) - 1))
````

**خروجی مورد انتظار:**

```text
آرایهٔ مرتب‌شده: [9, 12, 18, 25, 27, 31, 38, 43]
نمونهٔ دوم: [-5, -2, 0, 4, 4, 7]
```

**پیچیدگی زمانی:** در بهترین، متوسط و بدترین حالت $\Theta(n\log n)$ است.

**پیچیدگی فضایی:** فهرست کمکی در مجموع به اندازهٔ `n` عنصر فضا می‌خواهد؛ پس فضای کمکی $O(n)$ است. عمق پشتهٔ بازگشتی نیز $O(\log n)$ است.

**نکته:** این پیاده‌سازی آرایهٔ اصلی را تغییر می‌دهد و همان آرایهٔ مرتب‌شده را برمی‌گرداند.

## ۶. فاکتوریل (Factorial)

فاکتوریل عدد `n` حاصل‌ضرب اعداد صحیح مثبت از ۱ تا `n` است. برای نمونه:

$$

5!=5\times4\times3\times2\times1=120

$$

رابطهٔ ریاضی فاکتوریل:

$$

n!=n\times(n-1)!

$$

شرط پایان:

$$

0!=1

$$

### رابطهٔ زمان اجرا

در هر مرحله فقط یک بار تابع با ورودی `n-1` فراخوانی می‌شود و بقیهٔ عملیات زمان ثابت دارند:

$$

T(n)=T(n-1)+\Theta(1),\qquad T(0)=\Theta(1)

$$

این رابطه شکل استاندارد قضیهٔ مستر را ندارد؛ چون اندازهٔ ورودی هر بار یک واحد کم می‌شود.

### حل با درخت بازگشتی و گسترش رابطه

درخت این رابطه در واقع یک زنجیره است؛ در هر سطح فقط یک فراخوانی جدید داریم:

```text
T(n)
 |
T(n-1)
 |
T(n-2)
 |
 ...
 |
T(1)
 |
T(0)
```

تعداد سطح‌ها متناسب با `n` است و هزینهٔ هر سطح ثابت است. با گسترش رابطه نیز داریم:

$$

\begin{aligned}
T(n)&=T(n-1)+c\\
&=T(n-2)+2c\\
&=T(n-3)+3c\\
&\ \vdots\\
&=T(0)+nc
\end{aligned}

$$

پس:

$$

\boxed{T(n)=\Theta(n)}

$$

### شبه‌کد

```text
FACTORIAL(n)

    if n == 0
        return 1

    return n * FACTORIAL(n - 1)
```

### کد پایتون

ورودی تابع باید عدد صحیح نامنفی باشد.

````python
def factorial(n):
    if n < 0:
        raise ValueError("عدد باید نامنفی باشد.")

    if n == 0:
        return 1

    return n * factorial(n - 1)


print("5! =", factorial(5))
print("0! =", factorial(0))
````

**خروجی مورد انتظار:**

```text
5! = 120
0! = 1
```

**پیچیدگی زمانی:** $\Theta(n)$، چون تابع در مجموع متناسب با `n` بار فراخوانی می‌شود.

**پیچیدگی فضایی:** عمق پشتهٔ بازگشتی $O(n)$ است.

## ۷. فیبوناچی (Fibonacci)

در دنبالهٔ فیبوناچی، هر جمله از جمع دو جملهٔ قبلی به دست می‌آید:

$$

0,1,1,2,3,5,8,13,21,\ldots

$$

رابطه:

$$

F(n)=F(n-1)+F(n-2)

$$

شرط‌های اولیه:

$$

F(0)=0,\qquad F(1)=1

$$

### رابطهٔ زمان اجرا

در پیاده‌سازی بازگشتی ساده، برای محاسبهٔ هر جمله (به‌جز دو حالت پایه) دو فراخوانی دیگر انجام می‌شود:

$$

T(n)=T(n-1)+T(n-2)+\Theta(1)

$$

این رابطه با قضیهٔ مستر استاندارد حل نمی‌شود؛ چون اندازهٔ دو زیرمسئله متفاوت است.

### درخت بازگشتی برای F(5)

```text
F(5)
├── F(4)
│   ├── F(3)
│   │   ├── F(2)
│   │   │   ├── F(1)
│   │   │   └── F(0)
│   │   └── F(1)
│   └── F(2)
│       ├── F(1)
│       └── F(0)
└── F(3)
    ├── F(2)
    │   ├── F(1)
    │   └── F(0)
    └── F(1)
```

درخت نشان می‌دهد بعضی محاسبات، مانند `F(3)` و `F(2)`، چند بار تکرار می‌شوند.

### محاسبهٔ پیچیدگی از روی تعداد فراخوانی‌ها

اگر `C(n)` تعداد کل فراخوانی‌های تابع باشد، داریم:

$$

C(n)=C(n-1)+C(n-2)+1

$$

و برای دو حالت پایه:

$$

C(0)=1,\qquad C(1)=1

$$

با تعریف $D(n)=C(n)+1$، رابطه به شکل زیر درمی‌آید:

$$

D(n)=D(n-1)+D(n-2)

$$

این همان رابطهٔ فیبوناچی است. برای پیدا کردن نرخ رشد، معادلهٔ مشخصه را می‌نویسیم:

$$

r^2-r-1=0

$$

ریشه‌ها:

$$

r=\frac{1\pm\sqrt{5}}{2}

$$

ریشهٔ بزرگ‌تر:

$$

\varphi=\frac{1+\sqrt{5}}{2}\approx1.618

$$

پس تعداد فراخوانی‌ها و زمان اجرای نسخهٔ بازگشتی ساده از مرتبهٔ زیر است:

$$

\boxed{T(n)=\Theta(\varphi^n)=\Theta(1.618^n)}

$$

**اصلاح نکتهٔ تصویر جزوه:** برای رابطهٔ بازگشتی فیبوناچی، عبارت $O(2^{n/2})$ کران بالای درستی نیست. رشد دقیق‌تر نسخهٔ بازگشتی ساده $\Theta(1.618^n)$ است؛ $O(2^n)$ نیز یک کران بالای ساده‌تر اما غیردقیق‌تر است.

### شبه‌کد

```text
FIBONACCI(n)

    if n == 0
        return 0

    if n == 1
        return 1

    return FIBONACCI(n - 1) + FIBONACCI(n - 2)
```

### کد پایتون



````python
def fibonacci(n):
    if n < 0:
        raise ValueError("n باید نامنفی باشد.")

    if n == 0:
        return 0

    if n == 1:
        return 1

    return fibonacci(n - 1) + fibonacci(n - 2)


print("Fibonacci(6) =", fibonacci(6))
print("Fibonacci(8) =", fibonacci(8))
````

**خروجی مورد انتظار:**

```text
Fibonacci(6) = 8
Fibonacci(8) = 21
```

**پیچیدگی فضایی:** عمق پشتهٔ بازگشتی در یک مسیر حداکثر متناسب با `n` است؛ پس فضای پشته $O(n)$ است.

**نکته:** با روش ذخیرهٔ جواب‌های قبلی (Memoization) می‌توان زمان را به $O(n)$ کاهش داد؛ اما این دفترچه نسخهٔ بازگشتی ساده را بررسی می‌کند.

## ۸. برج هانوی (Tower of Hanoi)

در مسئلهٔ برج هانوی سه میله داریم:

- مبدأ (A)
- کمکی (B)
- مقصد (C)

تعدادی دیسک با اندازه‌های متفاوت روی میلهٔ مبدأ قرار دارند. هدف این است که همهٔ دیسک‌ها به میلهٔ مقصد منتقل شوند.

### قوانین مسئله

1. در هر حرکت فقط یک دیسک جابه‌جا می‌شود.
2. فقط دیسک بالایی یک میله را می‌توان برداشت.
3. دیسک بزرگ‌تر نباید روی دیسک کوچک‌تر قرار بگیرد.

### روش حل

برای انتقال `n` دیسک از مبدأ به مقصد:

1. ابتدا `n-1` دیسک را از مبدأ به میلهٔ کمکی منتقل می‌کنیم.
2. دیسک بزرگ‌تر را از مبدأ به مقصد می‌بریم.
3. سپس `n-1` دیسک را از میلهٔ کمکی به مقصد منتقل می‌کنیم.

### رابطهٔ بازگشتی

$$

T(n)=2T(n-1)+\Theta(1)

$$

این رابطه در فرم استاندارد قضیهٔ مستر نیست؛ چون اندازهٔ مسئله در هر مرحله یک واحد کم می‌شود.

### حل با درخت بازگشتی

هر فراخوانی، به‌جز حالت پایه، دو فراخوانی با اندازهٔ `n-1` ایجاد می‌کند و خودش یک حرکت انجام می‌دهد:

```text
                T(n)
              /      \\
          T(n-1)    T(n-1)
           /  \\      /  \\
       ...    ...  ...    ...
```

در سطح صفر یک گره، در سطح یک دو گره، در سطح دو چهار گره و به همین ترتیب داریم. آخرین سطح داخلی، سطح `n-1` است. هر گرهٔ داخلی یک حرکت انجام می‌دهد؛ پس تعداد حرکت‌ها برابر است با:

$$

M(n)=1+2+4+\cdots+2^{n-1}

$$

مجموع دنبالهٔ هندسی:

$$

M(n)=\frac{2^n-1}{2-1}=2^n-1

$$

پس:

$$

\boxed{T(n)=\Theta(2^n)}

$$

### رابطهٔ دقیق تعداد حرکت‌ها

$$

M(n)=2M(n-1)+1,\qquad M(0)=0

$$

با گسترش رابطه نیز به همان جواب می‌رسیم:

$$

M(n)=2^n-1

$$

مثلاً برای ۳ دیسک:

$$

M(3)=2^3-1=7

$$

### شبه‌کد

```text
HANOI(n, source, auxiliary, destination)

    if n == 0
        return

    HANOI(n - 1, source, destination, auxiliary)

    print source -> destination

    HANOI(n - 1, auxiliary, source, destination)
```

### کد پایتون

````python
def hanoi(n, source, auxiliary, destination):
    if n < 0:
        raise ValueError("تعداد دیسک‌ها نباید منفی باشد.")

    if n == 0:
        return

    hanoi(n - 1, source, destination, auxiliary)

    print(source, "->", destination)

    hanoi(n - 1, auxiliary, source, destination)


hanoi(3, "A", "B", "C")
````

**خروجی برای ۳ دیسک:**

```text
A -> C
A -> B
C -> B
A -> C
B -> A
B -> C
A -> C
```

تعداد حرکت‌ها دقیقاً ۷ است.

**پیچیدگی زمانی:** $\Theta(2^n)$، چون تعداد حرکت‌ها دقیقاً $2^n-1$ است.

**پیچیدگی فضایی:** عمق پشتهٔ بازگشتی `n` است؛ بنابراین فضای پشته $O(n)$ خواهد بود. این مقدار، فضای پشته است و با تعداد حرکت‌ها تفاوت دارد.

## ۹. جدول جمع‌بندی

| الگوریتم | رابطهٔ بازگشتی | روش حل | پیچیدگی زمانی |
|---|---|---|---|
| جست‌وجوی دودویی | $T(n)=T(n/2)+1$ | مستر، حالت دوم | $\Theta(\log n)$ |
| بیشینه‌یابی بازگشتی | $T(n)=2T(n/2)+1$ | مستر، حالت اول | $\Theta(n)$ |
| مرتب‌سازی ادغامی | $T(n)=2T(n/2)+n$ | مستر، حالت دوم | $\Theta(n\log n)$ |
| فاکتوریل | $T(n)=T(n-1)+1$ | درخت زنجیره‌ای / گسترش | $\Theta(n)$ |
| فیبوناچی ساده | $T(n)=T(n-1)+T(n-2)+1$ | درخت بازگشتی / معادلهٔ مشخصه | $\Theta(1.618^n)$ |
| برج هانوی | $T(n)=2T(n-1)+1$ | درخت بازگشتی / گسترش | $\Theta(2^n)$ |

### پیچیدگی فضایی الگوریتم‌ها

| الگوریتم | فضای کمکی یا پشتهٔ بازگشتی |
|---|---|
| جست‌وجوی دودویی بازگشتی | $O(\log n)$ |
| بیشینه‌یابی بازگشتی | $O(\log n)$ |
| مرتب‌سازی ادغامی | $O(n)$ فضای کمکی و $O(\log n)$ پشته |
| فاکتوریل بازگشتی | $O(n)$ |
| فیبوناچی بازگشتی ساده | $O(n)$ |
| برج هانوی | $O(n)$ پشته |

### چطور روش حل را تشخیص بدهیم؟

- اگر رابطه به شکل $T(n)=aT(n/b)+f(n)$ باشد، بررسی کن که قضیهٔ مستر قابل استفاده است یا نه.
- اگر رابطه مثل $T(n)=T(n-1)+c$ باشد، درخت بازگشتی آن یک زنجیره است و گسترش رابطه هم روش ساده‌ای برای حل است.
- اگر رابطه شامل دو زیرمسئله با اندازه‌های متفاوت، مثل $T(n-1)$ و $T(n-2)$، باشد، درخت بازگشتی و معادلهٔ مشخصه می‌توانند کمک کنند.
- اگر هر فراخوانی دو زیرمسئله با اندازهٔ `n-1` بسازد، مانند برج هانوی، تعداد گره‌ها به‌صورت هندسی افزایش پیدا می‌کند.
